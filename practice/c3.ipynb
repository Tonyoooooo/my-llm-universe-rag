```python
# 导入 Python 内置的正则表达式模块 re
import re

# re.compile(...)：把正则表达式预编译成一个规则对象，供后面匹配/替换使用
# r'...'：原始字符串（raw string），让里面的反斜杠 \u、\n 直接交给正则解析，不被 Python 字符串提前转义
# [^\u4e00-\u9fff]：\u4e00-\u9fff 是常用中文汉字的编码区间；方括号里开头加 ^ 表示“取反”，即匹配 1 个“非汉字字符”（如标点、字母、数字、空格）
# (\n)：匹配中间的换行符（括号表示第 1 个捕获分组）
# 整串合起来就是匹配：“非汉字 + 换行符 + 非汉字”（例如 "。\n”" 或 "a\nb"）
# re.DOTALL：正则修饰符，作用是让通配符 . 也能匹配换行符（本句正则里没用到 .，加不加效果一样）
pattern = re.compile(r'[^\u4e00-\u9fff](\n)[^\u4e00-\u9fff]', re.DOTALL)

# re.sub(规则, 替换方式, 原字符串)：在 pdf_page.page_content 里找出所有符合 pattern 的片段并替换，最后赋回给原属性
# lambda match: ...：匿名函数，每匹配到一处就会调用一次，参数 match 是当前匹配到的结果对象
# match.group(0)：取出这次匹配到的完整 3 个字符（如 "。\n”"）
# .replace('\n', '')：只把这 3 个字符中间的换行符删掉，保留两边的非汉字字符（变成 "。”"）
pdf_page.page_content = re.sub(pattern, lambda match: match.group(0).replace('\n', ''), pdf_page.page_content)

# 打印清洗后的这一页文本内容
print(pdf_page.page_content)
```

```python
import os
from dotenv import load_dotenv, find_dotenv

# 读取本地/项目的环境变量。
# find_dotenv()寻找并定位.env文件的路径
# load_dotenv()读取该.env文件，并将其中的环境变量加载到当前的运行环境中  
# 如果你设置的是全局的环境变量，这行代码则没有任何作用。
_ = load_dotenv(find_dotenv())

# 如果你需要通过代理端口访问，你需要如下配置
# os.environ['HTTPS_PROXY'] = 'http://127.0.0.1:7890'
# os.environ["HTTP_PROXY"] = 'http://127.0.0.1:7890'

# 获取folder_path下所有文件路径，储存在file_paths里
file_paths = []
folder_path = '../data_base/knowledge_db'
for root, dirs, files in os.walk(folder_path):
    for file in files:
        file_path = os.path.join(root, file)
        file_paths.append(file_path)
print(file_paths[:3])
```

```python
from langchain_community.document_loaders import PyMuPDFLoader
from langchain_community.document_loaders import TextLoader

# 遍历文件路径并把实例化的loader存放在loaders里
loaders = []

for file_path in file_paths:
    file_type = file_path.split('.')[-1]
    if file_type == 'pdf':
        loaders.append(PyMuPDFLoader(file_path))
    elif file_type == 'md':
        loaders.append(TextLoader(file_path))
texts = []

for loader in loaders: texts.extend(loader.load())

print(f"一共加载了 {len(texts)} 个文件")

text = texts[1]
print(f"每一个元素的类型：{type(text)}.", 
    f"该文档的描述性数据：{text.metadata}", 
    f"查看该文档的内容:\n{text.page_content[0:]}", 
    sep="\n------\n")
```
